```python
# 219
def print_arithmetic_operation(a, b):
    print(f"{a} + {b} = {a+b}")
    print(f"{a} - {b} = {a-b}")
    print(f"{a} * {b} = {a*b}")
    print(f"{a} / {b} = {a/b}")

print_arithmetic_operation(3, 4)
```

```python
# 220
def print_max(a, b, c):
    max_val = 0
    if a > max_val:
        max_val = a
    if b > max_val:
        max_val = b
    if c > max_val:
        max_val = c
    print(max_val)

print_max(3,5,8)
```

```python
# 233
def make_list(string):
    items = []
    for ch in string:
        items.append(ch)
    return items

def make_list2(string):
    return list(string)

print(make_list("abcd"))
print(make_list2("abcd"))
```

```python
# 234
def pickup_even(numbers):
    even = []
    for i in numbers:
        if i%2==0:
            even.append(i)
    return even

pickup_even([3, 4, 5, 6, 7, 8])
```

```python
# 235
def convert_int(string):
    string = string.replace(",", "")
    result = int(string)
    return result

convert_int("1,234,567")
```

```python
# 239
def 함수1(num) :
    return num + 4

def 함수2(num) :
    num = num + 2
    return 함수1(num)

c = 함수2(10)
print(c)
# 예측 결과: 16
```

```python
# 240
def 함수0(num) :
    return num * 2

def 함수1(num) :
    return 함수0(num + 2)

def 함수2(num) :
    num = num + 10
    return 함수1(num)

c = 함수2(2)
print(c)
# 예측 결과: 28
```
